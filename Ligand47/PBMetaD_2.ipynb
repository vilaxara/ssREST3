```python
def fe_regions(x, y):
    return np.stack([
        (x >= 0.4) & (x <  0.8) & (y >= 0) & (y <= 18),
        (x >= 0.8) & (x <= 2.0) & (y >= 0) & (y < 6),
        (x >= 0.8) & (x <= 2.0) & (y >= 6) & (y < 12),
        (x >= 0.8) & (x <= 2.0) & (y >= 12) & (y <= 18),
        (x > 2.0) & (x <= 2.4) & (y >= 0) & (y <= 18)
        
    ]).astype(int)
```
